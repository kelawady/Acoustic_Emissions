# AE Regime-Transition Analysis from Strain and Acoustic Emission

This notebook is a self-contained follow-on analysis that assumes the same data locations used in the earlier Colabs.

Goals:

1. **Identify the strain inflection point more accurately** by fitting two straight lines to strain vs time and optimizing the breakpoint.
2. **Match the inflection point to the first "large" AE event** using an energy threshold defined from the AE event energy distribution.
3. **Characterize AE events before vs after the transition** using timing, energy, spectral, and morphology features.
4. **Test earthquake-style acceleration diagnostics** using inverse-time / time-to-failure style fits on AE activity near the transition.
5. Optionally **recompute 4 AE archetypes** and compare cluster composition before and after the transition.

The notebook is conservative and interpretable: it tries simple analyses first, then adds a few stronger diagnostics.

## Build an event table with physically meaningful AE features

```python
FS_HZ = 2_000_000.0

def _rms(x, eps=1e-12):
    return float(np.sqrt(np.mean(x * x))) + eps

def _zcr_per_sample(x):
    return float(np.mean(np.signbit(x[1:]) != np.signbit(x[:-1]))) if len(x) >= 2 else np.nan

def _spec_features_abs_freq(x, fs_hz=FS_HZ):
    X = np.fft.rfft(x)
    mag = np.abs(X)
    if mag.size == 0:
        return np.nan, np.nan, np.nan
    mag[0] = 0.0
    f_hz = np.fft.rfftfreq(len(x), d=1.0 / fs_hz)
    mag_sum = float(np.sum(mag)) + 1e-12
    p = mag / mag_sum
    centroid_hz = float(np.sum(f_hz * mag) / mag_sum)
    bandwidth_hz = float(np.sqrt(np.sum(((f_hz - centroid_hz) ** 2) * mag) / mag_sum))
    entropy = float(-np.sum(p * np.log(p + 1e-12)))
    return centroid_hz, bandwidth_hz, entropy

def _env_rise_decay_fracs(x):
    env = np.abs(hilbert(x))
    N = len(env)
    if N == 0:
        return np.nan, np.nan
    pk = int(np.argmax(env))
    rise = float(pk / max(N - 1, 1))
    pk_env = float(env[pk])
    target = pk_env / np.e
    after = env[pk:]
    below = np.where(after <= target)[0]
    dec_idx = pk + int(below[0]) if below.size > 0 else N - 1
    decay = float((dec_idx - pk) / max(N - 1, 1))
    return rise, decay

rows = []
for i, ev in enumerate(events_sorted):
    x = np.asarray(ev.get("signal", []), dtype=float)
    x = x[np.isfinite(x)]
    if x.size < 8:
        continue

    rms = _rms(x)
    xn = x / rms
    energy = float(np.sum(x**2))
    peak_abs = float(np.max(np.abs(x)))
    zcr = _zcr_per_sample(xn)
    centroid_hz, bandwidth_hz, entropy = _spec_features_abs_freq(xn, FS_HZ)
    rise_frac, decay_frac = _env_rise_decay_fracs(xn)

    start_t = float(ev.get("global_start_time_s", np.nan))
    dur_s = float(ev.get("duration_s", x.size / FS_HZ))
    end_t = start_t + dur_s

    rows.append({
        "event_idx": i,
        "t_start": start_t,
        "t_end": end_t,
        "duration_s": dur_s,
        "duration_us": dur_s * 1e6,
        "energy": energy,
        "log10_energy": np.log10(energy + 1e-12),
        "rms": rms,
        "peak_abs": peak_abs,
        "kurtosis": float(kurtosis(xn, fisher=False, bias=False)),
        "zcr_cross_per_sample": zcr,
        "zcr_hz": zcr * FS_HZ if np.isfinite(zcr) else np.nan,
        "spec_centroid_khz": centroid_hz / 1e3,
        "spec_bandwidth_khz": bandwidth_hz / 1e3,
        "spec_entropy": entropy,
        "rise_frac": rise_frac,
        "decay_frac": decay_frac,
    })

events_df = pd.DataFrame(rows).sort_values("t_start").reset_index(drop=True)
events_df.head()
```

## Fit a two-line piecewise linear model to strain vs time

We scan possible breakpoints and pick the breakpoint with the smallest total squared error from fitting a straight line on each side.

This is simple, robust, and easy to interpret.

```python
def piecewise_two_line_fit(t, y, min_frac=0.15):
    t = np.asarray(t, dtype=float)
    y = np.asarray(y, dtype=float)
    n = len(t)
    lo = int(max(5, np.floor(min_frac * n)))
    hi = int(min(n - 5, np.ceil((1 - min_frac) * n)))

    best = None
    out = []

    for bp in range(lo, hi):
        t1, y1 = t[:bp], y[:bp]
        t2, y2 = t[bp:], y[bp:]

        p1 = np.polyfit(t1, y1, 1)
        p2 = np.polyfit(t2, y2, 1)

        yhat1 = np.polyval(p1, t1)
        yhat2 = np.polyval(p2, t2)

        sse1 = float(np.sum((y1 - yhat1)**2))
        sse2 = float(np.sum((y2 - yhat2)**2))
        sse = sse1 + sse2

        cand = {
            "bp_idx": bp,
            "bp_time": float(t[bp]),
            "p1": p1,
            "p2": p2,
            "sse": sse,
            "slope1": float(p1[0]),
            "slope2": float(p2[0]),
        }
        out.append(cand)

        if (best is None) or (sse < best["sse"]):
            best = cand

    return best, pd.DataFrame(out)

fit_best, fit_scan = piecewise_two_line_fit(t_mech, strain, min_frac=0.15)
fit_best
```

```python
# Plot the best two-line fit and breakpoint
FONT_TITLE = 16
FONT_LABEL = 14
FONT_TICKS = 12

bp_idx = fit_best["bp_idx"]
bp_time = fit_best["bp_time"]

plt.figure(figsize=(10, 5))
plt.plot(t_mech, strain, label="Strain", linewidth=2)

plt.plot(t_mech[:bp_idx], np.polyval(fit_best["p1"], t_mech[:bp_idx]),
         "--", linewidth=2, label=f"Fit before (slope={fit_best['slope1']:.4g})")
plt.plot(t_mech[bp_idx:], np.polyval(fit_best["p2"], t_mech[bp_idx:]),
         "--", linewidth=2, label=f"Fit after (slope={fit_best['slope2']:.4g})")

plt.axvline(bp_time, color="k", linestyle=":", linewidth=2, label=f"Estimated inflection = {bp_time:.2f} s")
plt.title("Strain vs Time with Optimized Two-Line Breakpoint", fontsize=FONT_TITLE)
plt.xlabel("Time (s)", fontsize=FONT_LABEL)
plt.ylabel("Strain", fontsize=FONT_LABEL)
plt.xticks(fontsize=FONT_TICKS)
plt.yticks(fontsize=FONT_TICKS)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()

print(f"Estimated strain inflection time: {bp_time:.4f} s")
print(f"Pre-break slope:  {fit_best['slope1']:.6g}")
print(f"Post-break slope: {fit_best['slope2']:.6g}")
print(f"Slope ratio (after/before): {fit_best['slope2']/fit_best['slope1']:.3f}")
```

## AE activity vs time around the transition

We compute rolling summaries on a common time grid:
- event rate
- AE energy rate
- median spectral centroid
- median entropy

This helps show whether activity accelerates into the transition.

```python
# Build time-grid rolling summaries
GRID_DT = 0.5      # seconds
WINDOW = 5.0       # rolling window in seconds

t_grid = np.arange(float(events_df["t_start"].min()), float(events_df["t_end"].max()) + GRID_DT, GRID_DT)

def window_stats(center_t, window_s, df):
    lo = center_t - window_s/2
    hi = center_t + window_s/2
    w = df[(df["t_start"] >= lo) & (df["t_start"] < hi)]
    n = len(w)
    duration = max(window_s, 1e-12)
    return {
        "event_rate_hz": n / duration,
        "energy_rate": float(w["energy"].sum() / duration),
        "median_log10_energy": float(w["log10_energy"].median()) if n else np.nan,
        "median_centroid_khz": float(w["spec_centroid_khz"].median()) if n else np.nan,
        "median_entropy": float(w["spec_entropy"].median()) if n else np.nan,
    }

grid_rows = []
for tg in t_grid:
    d = window_stats(tg, WINDOW, events_df)
    d["t"] = tg
    grid_rows.append(d)

grid_df = pd.DataFrame(grid_rows).sort_values("t")
grid_df.head()
```

```python
fig, axes = plt.subplots(4, 1, figsize=(10, 10), sharex=True)

series = [
    ("event_rate_hz", "Event rate (events/s)"),
    ("energy_rate", "AE energy rate"),
    ("median_centroid_khz", "Rolling median centroid (kHz)"),
    ("median_entropy", "Rolling median spectral entropy"),
]

for ax, (col, label) in zip(axes, series):
    ax.plot(grid_df["t"], grid_df[col], linewidth=2)
    ax.axvline(bp_time, color="k", linestyle="--", linewidth=1.7, label="strain breakpoint")
    ax.set_ylabel(label)
    ax.legend(fontsize=9)

axes[-1].set_xlabel("Time (s)")
plt.suptitle("Rolling AE activity summaries around the inferred transition", y=0.995)
plt.tight_layout()
plt.show()
```

## Simple transition predictor from AE activity

This is not a full forecasting model. It just asks:
**Can short-window AE summaries tell us whether we are close to the transition?**

Target = 1 if time is within `HORIZON_S` before the estimated transition.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score, StratifiedKFold

HORIZON_S = 5.0

pred_df = grid_df.copy()
pred_df["time_to_transition"] = bp_time - pred_df["t"]
pred_df = pred_df[(pred_df["time_to_transition"] >= 0)]  # only before transition
pred_df["target"] = (pred_df["time_to_transition"] <= HORIZON_S).astype(int)

feat_cols_pred = ["event_rate_hz", "energy_rate", "median_log10_energy", "median_centroid_khz", "median_entropy"]
pred_use = pred_df.dropna(subset=feat_cols_pred + ["target"]).copy()

X = pred_use[feat_cols_pred].values
y = pred_use["target"].values

if len(np.unique(y)) >= 2 and len(pred_use) >= 20:
    scaler_pred = StandardScaler()
    Xz = scaler_pred.fit_transform(X)
    clf = LogisticRegression(max_iter=2000)
    n_splits = min(5, int(np.bincount(y).min())) if np.bincount(y).min() >= 2 else 0
    if n_splits >= 2:
        cv = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=0)
        aucs = cross_val_score(clf, Xz, y, cv=cv, scoring="roc_auc")
        print("Transition-warning ROC AUCs:", np.round(aucs, 3))
        print("Mean ROC AUC:", float(np.mean(aucs)))
    else:
        print("Not enough positive windows for stable cross-validation.")
else:
    print("Not enough positive/negative windows to run the simple predictor robustly.")
```

## Earthquake-style inverse-time / time-to-failure acceleration analysis

A simple diagnostic from failure forecasting is to test whether AE activity accelerates as

\[
r(t) = a + \frac{b}{(t_c - t)^p}
\]

as time approaches a critical time `t_c`.

Here:
- `t_c` is initialized to the estimated strain breakpoint,
- `r(t)` is a rolling AE activity measure,
- we fit this model on the **pre-transition** interval only.

This is exploratory, but it is a useful way to test whether the AE dynamics show accelerating approach to a regime transition.

```python
def inverse_time_model(t, a, b, p, tc):
    eps = 1e-6
    return a + b / np.maximum(tc - t, eps)**p

# choose a pre-transition fitting window
PRE_WINDOW_S = 25.0
fit_lo = max(float(grid_df["t"].min()), bp_time - PRE_WINDOW_S)
fit_hi = bp_time - 0.5  # keep away from singular endpoint

fit_mask = (grid_df["t"] >= fit_lo) & (grid_df["t"] <= fit_hi)
fit_t = grid_df.loc[fit_mask, "t"].values
fit_r = grid_df.loc[fit_mask, "energy_rate"].values

# remove nans
m = np.isfinite(fit_t) & np.isfinite(fit_r)
fit_t = fit_t[m]
fit_r = fit_r[m]

print("Fit points:", len(fit_t), "window:", fit_lo, "to", fit_hi)

inv_fit = None
if len(fit_t) >= 8:
    try:
        p0 = [np.nanmin(fit_r), max(np.nanmax(fit_r)-np.nanmin(fit_r), 1e-3), 0.5, bp_time]
        bounds = ([-np.inf, 0.0, 0.01, bp_time - 5.0],
                  [ np.inf, np.inf, 3.0, bp_time + 5.0])
        popt, pcov = curve_fit(inverse_time_model, fit_t, fit_r, p0=p0, bounds=bounds, maxfev=20000)
        inv_fit = popt
        print("Fitted inverse-time params [a, b, p, tc]:", popt)
        print(f"Estimated critical time tc = {popt[3]:.4f} s; strain breakpoint = {bp_time:.4f} s")
    except Exception as e:
        print("Inverse-time fit failed:", e)
else:
    print("Not enough points for inverse-time fit.")
```

```python
# Plot inverse-time fit if available
plt.figure(figsize=(9, 4.8))
plt.plot(grid_df["t"], grid_df["energy_rate"], linewidth=2, label="Rolling AE energy rate")
plt.axvline(bp_time, color="k", linestyle="--", linewidth=1.7, label=f"Strain breakpoint {bp_time:.2f}s")

if inv_fit is not None:
    t_plot = np.linspace(fit_lo, fit_hi, 200)
    r_plot = inverse_time_model(t_plot, *inv_fit)
    plt.plot(t_plot, r_plot, "r--", linewidth=2,
             label=f"Inverse-time fit (p={inv_fit[2]:.2f}, tc={inv_fit[3]:.2f}s)")

plt.xlabel("Time (s)")
plt.ylabel("AE energy rate")
plt.title("Earthquake-style inverse-time acceleration fit")
plt.legend(fontsize=9)
plt.tight_layout()
plt.show()
```

## Alternative acceleration check in log coordinates

If acceleration is approximately power-law in time-to-failure, then plotting activity vs `(t_c - t)` on log axes can reveal a near-linear trend.

```python
if inv_fit is not None:
    tc_use = inv_fit[3]
else:
    tc_use = bp_time

tf = tc_use - fit_t
rr = fit_r

mask = (tf > 0) & (rr > 0) & np.isfinite(tf) & np.isfinite(rr)
tf = tf[mask]
rr = rr[mask]

if len(tf) >= 5:
    lx = np.log10(tf)
    ly = np.log10(rr)
    m, b = np.polyfit(lx, ly, 1)

    xfit = np.logspace(np.log10(tf.min()), np.log10(tf.max()), 200)
    yfit = 10**b * xfit**m

    plt.figure(figsize=(6.5, 5))
    plt.scatter(tf, rr, s=35, label="Pre-transition windows")
    plt.plot(xfit, yfit, "--", linewidth=2, label=f"log10(r) = {m:.3f} log10(tc - t) + {b:.3f}")
    plt.xscale("log")
    plt.yscale("log")
    plt.xlabel("Time-to-transition (tc - t) [s]")
    plt.ylabel("AE energy rate")
    plt.title("Log-log acceleration diagnostic")
    plt.legend(fontsize=9)
    plt.tight_layout()
    plt.show()

    print(f"log10(r) = {m:.4f} log10(tc - t) + {b:.4f}")
else:
    print("Not enough positive points for log-log acceleration diagnostic.")
```

## Optional: recompute 4 AE archetypes and compare cluster composition before vs after

This is slower, but it lets you test whether the inferred deformation transition changes the mixture of AE archetypes.

```python
RUN_ARCHETYPES = True  # set False to skip
```

```python
if RUN_ARCHETYPES:
    def _wavelet_feats(x, wavelet="db4", level=3):
        if not _HAS_PYWT:
            return None
        coeffs = pywt.wavedec(x, wavelet, level=level)
        A3, D3, D2, D1 = coeffs[0], coeffs[1], coeffs[2], coeffs[3]
        EA3 = float(np.sum(np.asarray(A3)**2))
        ED3 = float(np.sum(np.asarray(D3)**2))
        ED2 = float(np.sum(np.asarray(D2)**2))
        ED1 = float(np.sum(np.asarray(D1)**2))
        Etot = EA3 + ED3 + ED2 + ED1 + 1e-12
        fr = np.array([ED1, ED2, ED3, EA3], dtype=float) / Etot
        went = float(-np.sum(fr * np.log(fr + 1e-12)))
        return list(fr) + [went, float(fr[0])]

    feat_rows = []
    keep_idx = []
    for i, ev in enumerate(events_sorted):
        x = np.asarray(ev.get("signal", []), dtype=float)
        x = x[np.isfinite(x)]
        if x.size < 32:
            continue
        rms = _rms(x)
        xn = x / rms
        centroid_hz, bandwidth_hz, entropy = _spec_features_abs_freq(xn, FS_HZ)
        rise_frac, decay_frac = _env_rise_decay_fracs(xn)
        zcr_ps = _zcr_per_sample(xn)
        row = [
            np.log10(float(ev.get("duration_s", x.size / FS_HZ)) + 1e-12),
            rise_frac,
            decay_frac,
            float(kurtosis(xn, fisher=False, bias=False)),
            zcr_ps * FS_HZ,
            centroid_hz / 1e3,
            bandwidth_hz / 1e3,
            entropy,
        ]
        wf = _wavelet_feats(xn)
        if wf is not None:
            row.extend(wf)
        feat_rows.append(row)
        keep_idx.append(i)

    X_event = np.asarray(feat_rows, dtype=float)
    Xz = StandardScaler().fit_transform(X_event)

    Ks = list(range(2, 7))
    bics = []
    gmms = []
    for k in Ks:
        gmm = GaussianMixture(n_components=k, covariance_type="full", random_state=0, n_init=10)
        gmm.fit(Xz)
        bics.append(gmm.bic(Xz))
        gmms.append(gmm)
    best_i = int(np.argmin(bics))
    best_k = Ks[best_i]
    best_gmm = gmms[best_i]
    labels = best_gmm.predict(Xz)

    arch_df = events_df.iloc[keep_idx].copy().reset_index(drop=True)
    arch_df["cluster"] = labels
    arch_df["regime"] = np.where(arch_df["t_start"] < bp_time, "before", "after")

    print("Best GMM k by BIC:", best_k)
    display(pd.crosstab(arch_df["cluster"], arch_df["regime"], normalize="columns"))

    frac_tab = pd.crosstab(arch_df["cluster"], arch_df["regime"], normalize="columns")
    frac_tab.plot(kind="bar", figsize=(8,4.5))
    plt.ylabel("Fraction of events")
    plt.title("Archetype composition before vs after transition")
    plt.tight_layout()
    plt.show()
```

## What to look for in the output

Good evidence that AE predicts the onset of increasing strain would look like one or more of these:

- The first large AE event occurs very close to the optimized strain breakpoint.
- AE event rate and/or AE energy rate accelerate into the breakpoint.
- Before-vs-after event populations differ substantially in energy, timing, centroid, or archetype composition.
- The inverse-time fit estimates a critical time close to the strain breakpoint.
- A short-window predictor can identify whether the system is within a few seconds of the transition.

If these hold consistently across experiments, then the next step would be a multi-run analysis that treats the inferred transition time as a target and tests generalization across specimens / loading runs.
